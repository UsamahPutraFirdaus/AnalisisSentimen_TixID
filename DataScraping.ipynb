```python
from google_play_scraper import app, reviews, Sort
import pandas as pd
```

```python
datascraping, _ = reviews(
  'id.tix.android',
  lang='id',
  country='id',
  sort = Sort.MOST_RELEVANT,
  count = 30000
)

reviews_df = pd.DataFrame(datascraping)
reviews_df
```

```python
# save to csv
reviews_df.to_csv('tixid_reviews.csv', index=False)
```
