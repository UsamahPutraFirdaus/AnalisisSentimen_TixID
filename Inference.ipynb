```python
import numpy as np
from tensorflow.keras.models import load_model
from tensorflow.keras.preprocessing.text import Tokenizer
from tensorflow.keras.preprocessing.sequence import pad_sequences
import pickle
import re

# Fungsi preprocessing
def preprocess_text(text):
    text = text.lower()
    text = re.sub(r'[^a-zA-Z\s]', '', text)
    text = re.sub(r'\s+', ' ', text).strip()
    return text

# ====== 1. Load Model dan Tokenizer ======
model = load_model('model_bilstm.keras')

with open('tokenizer_bilstm.pkl', 'rb') as handle:
    tokenizer = pickle.load(handle)

# ====== 2. Input Manual ======
test_input = "aplikasi tix id mempermudah pengunjung bioskop praktis simpel menonton bioskop mengantri tiket pesan rumah menghemat lihat film terbaru tayang mudah memilih bioskop kesukaan jam tayang memilih kursi mudah praktis menyenangkan tix id"

test_input_preprocessed = preprocess_text(test_input)
print(f'Teks setelah preprocessing: {test_input_preprocessed}')

# ====== 3. Tokenisasi & Padding ======
max_len = 100
test_seq = tokenizer.texts_to_sequences([test_input_preprocessed])
test_pad = pad_sequences(test_seq, maxlen=max_len, padding='post')

# ====== 4. Prediksi ======
pred_proba = model.predict(test_pad)
pred_label = np.argmax(pred_proba, axis=1)[0]

# ====== 5. Mapping Label ======
label_map = {
    0: 'Negatif',
    1: 'Netral',
    2: 'Positif'
}
sentimen = label_map.get(pred_label, 'Label tidak dikenali')

print(f'Hasil prediksi: {sentimen}')
```
